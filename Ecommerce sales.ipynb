```python
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df=df.apply(lambda x: x.str.strip() if x.dtype =='str' else x)
df.head()
```

```python
df['Buyer_Name'] = df['Buyer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
```

```python
df = df[df['Email'].str.contains('@',na=False)]
```

```python
df['Purchase_Date']=pd.to_datetime(df['Purchase_Date'],errors='coerce')
df['Delivered_Date']=pd.to_datetime(df['Delivered_Date'],errors='coerce')
```

```python
cols=['Quantity','Selling_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
    
```

```python
df = df[df['Quantity']>0]
```

```python
df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('Unknown')
df['Delivered_Date']=df['Delivered_Date'].fillna(df['Purchase_Date'])
```

```python
df.head()
```

```python
df['Sales']=df['Quantity']*df['Selling_Price']
```

```python
df
```

```python
df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
df['Profit']=df['Net_Amount']*0.20
```

```python
df['Month']=df['Purchase_Date'].dt.month_name()
```

```python
df['Year']=df['Purchase_Date'].dt.year
```

```python
df['Week_Day']=df['Purchase_Date'].dt.day_name()
```
